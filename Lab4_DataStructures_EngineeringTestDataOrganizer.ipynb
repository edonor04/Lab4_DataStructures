```python
LAST_NAME = "Donor"
STUDENT_ID = "TUPM-26-7923"
FAVORITE_ARTIST = "Paramore"

seed_digit = int(STUDENT_ID[-1])
id_checksum = sum(int(d) for d in STUDENT_ID if d.isdigit())
vector_dim = len(LAST_NAME)

sys_config = {
    "operator": LAST_NAME,
    "auth_id": STUDENT_ID,
    "artist": FAVORITE_ARTIST,
    "base_seed": seed_digit,
    "checksum": id_checksum,
    "vector_dim": vector_dim,
    "status": "INITIALIZED"
}

print("=== SYSTEM CONFIGURATION ===")
for key, value in sys_config.items():
    print(f"{key.upper()}: {value}")
```

```python
print("\n=== 1. GENERATED TEST DATA ===")
 
sources = [f"{LAST_NAME}_T{n}" for n in range(1, 6)]
categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]
 
artist_positions = [pos for pos in range(1, len(FAVORITE_ARTIST) + 1)]
print(f"Artist Positions: {artist_positions}")
 
values = [(pos * sys_config["base_seed"]) + sys_config["checksum"] for pos in artist_positions]
print(f"Generated Values: {values}")
 
test_records = []
 
for i in range(5):
    test_records.append((sources[i], categories[i], values[i]))
 
test_records.append((sources[0], categories[1], values[5]))
test_records.append((sources[2], categories[3], values[6]))
 
test_records.append((sources[0], categories[0], values[7]))
 
print(f"\n{'SOURCE':<12}{'CATEGORY':<14}{'VALUE'}")
for source, category, value in test_records:
    print(f"{source:<12}{category:<14}{value}")
```

```python
print("\n=== 2. ORGANIZED TEST DATA ===")
 
organized_data = {}
for source, category, value in test_records:
    key = (source, category)
    if key not in organized_data:
        organized_data[key] = []
    organized_data[key].append(value)
 
print("By Source and Category:")
for key, value_list in organized_data.items():
    print(f"{key[0]} | {key[1]}: {value_list}")
 
category_data = {}
for source, category, value in test_records:
    if category not in category_data:
        category_data[category] = []
    category_data[category].append(value)
 
source_data = {}
for source, category, value in test_records:
    if source not in source_data:
        source_data[source] = []
    source_data[source].append(value)
 
print("\nBy Category:")
for category, value_list in category_data.items():
    print(f"{category}: {value_list}")
```

```python
print("\n=== 3. REPEATED / DISTINCT DATA ===")
 
repeated_records = []
for key, value_list in organized_data.items():
    if len(value_list) > 1:
        repeated_records.append(key)
 
print(f"Repeated Source/Category Pairs: {repeated_records}")
 
all_values = [value for source, category, value in test_records]
distinct_values = set(all_values)
distinct_sources = set([source for source, category, value in test_records])
distinct_categories = set([category for source, category, value in test_records])
distinct_pairs = set(organized_data.keys())
 
print(f"Total Records: {len(test_records)}")
print(f"Distinct Pairs: {len(distinct_pairs)}")
print(f"Distinct Sources: {len(distinct_sources)}")
print(f"Distinct Categories: {len(distinct_categories)}")
print(f"Distinct Values: {len(distinct_values)} of {len(all_values)}")
```

```python
print("\n=== 4. NUMERICAL AND CATEGORICAL SUMMARIES ===")
 
category_summary = {}
for category, value_list in category_data.items():
    category_summary[category] = {
        "count": len(value_list),
        "minimum": min(value_list),
        "maximum": max(value_list),
        "total": sum(value_list),
        "average": round(sum(value_list) / len(value_list), 2)
    }
 
print("Category Summary (numerical):")
for category, summary in category_summary.items():
    print(f"{category}: {summary}")
 
source_summary = {}
for source, value_list in source_data.items():
    source_summary[source] = {
        "count": len(value_list),
        "total": sum(value_list),
        "average": round(sum(value_list) / len(value_list), 2)
    }
 
print("\nSource Summary (numerical):")
for source, summary in source_summary.items():
    print(f"{source}: {summary}")
 
print("\nRecord Count per Category (categorical):")
for category, summary in category_summary.items():
    print(f"{category}: {summary['count']} record(s)")
```

```python
print("\n=== 5. PROCESSED RESULTS ===")
 
unique_records = []
seen_pairs = set()
for source, category, value in test_records:
    if (source, category) not in seen_pairs:
        unique_records.append((source, category, value))
        seen_pairs.add((source, category))
 
print(f"Records Before Removing Repeats: {len(test_records)}")
print(f"Records After Removing Repeats: {len(unique_records)}")
 
pairs_a = set([(s, c) for s, c, v in test_records[:5]])
pairs_b = set([(s, c) for s, c, v in test_records[5:]])
combined_pairs = pairs_a.union(pairs_b)
common_pairs = pairs_a.intersection(pairs_b)
print(f"Pairs in first 5 records and later records (Common): {common_pairs}")
print(f"Union of Pairs: {len(combined_pairs)} pairs")
```

```python
print("\n=== 6. ORDERED RESULTS ===")
 
category_ranking = []
for category, summary in category_summary.items():
    category_ranking.append((summary["average"], category))
category_ranking.sort(reverse=True)
 
print("Categories by Average (Highest to Lowest):")
for average, category in category_ranking:
    print(f"{category}: {average}")
 
sorted_values = sorted(all_values)
print(f"\nAll Values Sorted: {sorted_values}")
 
value_ranking = []
for source, category, value in unique_records:
    value_ranking.append((value, source, category))
value_ranking.sort()
 
print("\nUnique Records by Value (Lowest to Highest):")
for value, source, category in value_ranking:
    print(f"{value} -> {source} | {category}")
```

```python
print("\n=== 7. FINAL SUMMARY ===")
 
count_ranking = []
for category, summary in category_summary.items():
    count_ranking.append((summary["count"], category))
count_ranking.sort(reverse=True)
 
print("=== FINAL ORGANIZED AND SUMMARIZED TEST DATASET ===")
print(f"Operator: {sys_config['operator']}")
print(f"Total Records: {len(test_records)}")
print(f"Unique Records: {len(unique_records)}")
print(f"Repeated Pairs: {repeated_records}")
print(f"Highest Value: {max(all_values)}")
print(f"Lowest Value: {min(all_values)}")
print(f"Overall Average: {round(sum(all_values) / len(all_values), 2)}")
print(f"Category with Highest Average: {category_ranking[0][1]} ({category_ranking[0][0]})")
print(f"Category with Most Records: {count_ranking[0][1]} ({count_ranking[0][0]} records)")
```
