```python
LAST_NAME = "Donor"
STUDENT_ID = "TUPM-26-7923"
FAVORITE_ARTIST = "Paramore"
 
seed_digit = int(STUDENT_ID[-1])
id_checksum = sum(int(d) for d in STUDENT_ID if d.isdigit())
vector_dim = len(LAST_NAME)
 
sys_config = {
    "operator": LAST_NAME,
    "auth_id": STUDENT_ID,
    "artist": FAVORITE_ARTIST,
    "base_seed": seed_digit,
    "checksum": id_checksum,
    "vector_dim": vector_dim,
    "status": "INITIALIZED"
}
 
print("=== SYSTEM CONFIGURATION ===")
for key, value in sys_config.items():
    print(f"{key.upper()}: {value}")
 
# Simple function that gives the transaction category based on its value
def get_transaction_category(value):
    if value >= 1000:
        return "LARGE"
    elif value >= 300:
        return "MEDIUM"
    else:
        return "SMALL"
```

```python
print("\n=== 1. GENERATED TRANSACTION RECORDS ===")
 
customers = [f"{LAST_NAME}_C{n}" for n in range(1, 6)]
products = ["NOTEBOOK", "PEN", "USB", "HEADPHONES", "CALCULATOR"]

prices = {
    "NOTEBOOK": 45,
    "PEN": 15,
    "USB": 350,
    "HEADPHONES": 800,
    "CALCULATOR": 500
}
 
print("Product Prices:")
for product, price in prices.items():
    print(f"{product}: {price}")
 
artist_positions = [pos for pos in range(1, len(FAVORITE_ARTIST) + 1)]
print(f"\nArtist Positions: {artist_positions}")
 
quantities = [((pos * sys_config["base_seed"]) % 5) + 1 for pos in artist_positions]
print(f"Generated Quantities: {quantities}")
 
raw_transactions = []
 
for i in range(5):
    raw_transactions.append((customers[i], products[i], quantities[i]))
 
raw_transactions.append((customers[0], products[0], quantities[5]))
 
raw_transactions.append((customers[1], products[2], quantities[6]))
raw_transactions.append((customers[2], products[1], quantities[7]))
 
print(f"\n{'CUSTOMER':<12}{'PRODUCT':<14}{'QTY'}")
for customer, product, quantity in raw_transactions:
    print(f"{customer:<12}{product:<14}{quantity}")
```

```python
print("\n=== 2. ORGANIZED TRANSACTION DATA ===")
 
transaction_list = []
transaction_number = 1
 
for customer, product, quantity in raw_transactions:
    price = prices[product]
    total_value = quantity * price
    transaction = {
        "id": transaction_number,
        "customer": customer,
        "product": product,
        "quantity": quantity,
        "price": price,
        "value": total_value,
        "category": get_transaction_category(total_value)
    }
    transaction_list.append(transaction)
    transaction_number += 1
 
print(f"{'ID':<5}{'CUSTOMER':<12}{'PRODUCT':<14}{'QTY':<6}{'PRICE':<8}{'VALUE'}")
for t in transaction_list:
    print(f"{t['id']:<5}{t['customer']:<12}{t['product']:<14}{t['quantity']:<6}{t['price']:<8}{t['value']}")
```

```python
print("\n=== 3. REPEATED/DISTINCT TRANSACTIONS ===")
 
pair_count = {}
for t in transaction_list:
    pair = (t["customer"], t["product"])
    if pair not in pair_count:
        pair_count[pair] = 0
    pair_count[pair] += 1
 
repeated_purchases = []
for pair, count in pair_count.items():
    if count > 1:
        repeated_purchases.append((pair, count))
 
print(f"Repeated Purchases (customer, product, times): {repeated_purchases}")
 
distinct_customers = set([t["customer"] for t in transaction_list])
distinct_products = set([t["product"] for t in transaction_list])
distinct_pairs = set(pair_count.keys())
 
print(f"Distinct Customers: {len(distinct_customers)}")
print(f"Distinct Products: {len(distinct_products)}")
print(f"Distinct Customer-Product Pairs: {len(distinct_pairs)} of {len(transaction_list)} transactions")
```

```python
print("\n=== 4. TRANSACTION VALUES ===")
 
for t in transaction_list:
    print(f"Transaction {t['id']}: {t['quantity']} x {t['price']} = {t['value']} ({t['category']})")
 
overall_total = sum([t["value"] for t in transaction_list])
total_quantity = sum([t["quantity"] for t in transaction_list])
print(f"\nOverall Transaction Value: {overall_total}")
print(f"Total Items Sold: {total_quantity}")
```

```python
print("\n=== 5. CUSTOMER SUMMARIES ===")
 
customer_summary = {}
for t in transaction_list:
    name = t["customer"]
    if name not in customer_summary:
        customer_summary[name] = {"transactions": 0, "items": 0, "total_value": 0}
    customer_summary[name]["transactions"] += 1
    customer_summary[name]["items"] += t["quantity"]
    customer_summary[name]["total_value"] += t["value"]
 
for name, summary in customer_summary.items():
    print(f"{name}: {summary}")
```

```python
print("\n=== 6. PRODUCT SUMMARIES ===")
 
product_summary = {}
for t in transaction_list:
    name = t["product"]
    if name not in product_summary:
        product_summary[name] = {"transactions": 0, "items": 0, "total_value": 0}
    product_summary[name]["transactions"] += 1
    product_summary[name]["items"] += t["quantity"]
    product_summary[name]["total_value"] += t["value"]
 
for name, summary in product_summary.items():
    print(f"{name}: {summary}")
```

```python
print("\n=== 7. ORDERED RESULTS ===")
 
customer_ranking = []
for name, summary in customer_summary.items():
    customer_ranking.append((summary["total_value"], name))
customer_ranking.sort(reverse=True)
 
print("Customers by Total Value (Highest to Lowest):")
rank = 1
for total, name in customer_ranking:
    print(f"{rank}. {name}: {total}")
    rank += 1
 
product_ranking = []
for name, summary in product_summary.items():
    product_ranking.append((summary["total_value"], name))
product_ranking.sort(reverse=True)
 
print("\nProducts by Total Value (Highest to Lowest):")
rank = 1
for total, name in product_ranking:
    print(f"{rank}. {name}: {total}")
    rank += 1
 
transaction_ranking = []
for t in transaction_list:
    transaction_ranking.append((t["value"], t["id"]))
transaction_ranking.sort(reverse=True)
 
print("\nTransactions by Value (Highest to Lowest):")
for value, transaction_id in transaction_ranking:
    print(f"Transaction {transaction_id}: {value}")
```

```python
print("\n=== 8. OVERALL TRANSACTION SUMMARY ===")
 
category_counts = {}
for t in transaction_list:
    if t["category"] not in category_counts:
        category_counts[t["category"]] = 0
    category_counts[t["category"]] += 1
 
print("=== FINAL CONSOLIDATED TRANSACTION REPORT ===")
print(f"Operator: {sys_config['operator']}")
print(f"Total Transactions: {len(transaction_list)}")
print(f"Total Items Sold: {total_quantity}")
print(f"Overall Transaction Value: {overall_total}")
print(f"Average Transaction Value: {round(overall_total / len(transaction_list), 2)}")
print(f"Repeated Purchases: {repeated_purchases}")
print(f"Transaction Categories: {category_counts}")
print(f"Top Customer: {customer_ranking[0][1]} ({customer_ranking[0][0]})")
print(f"Top Product: {product_ranking[0][1]} ({product_ranking[0][0]})")
print(f"Lowest Product: {product_ranking[-1][1]} ({product_ranking[-1][0]})")
```
