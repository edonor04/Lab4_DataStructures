```python
LAST_NAME = "Donor"
STUDENT_ID = "TUPM-26-7923"
FAVORITE_ARTIST = "Paramore"
 
seed_digit = int(STUDENT_ID[-1])
id_checksum = sum(int(d) for d in STUDENT_ID if d.isdigit())
vector_dim = len(LAST_NAME)
 
sys_config = {
    "operator": LAST_NAME,
    "auth_id": STUDENT_ID,
    "artist": FAVORITE_ARTIST,
    "base_seed": seed_digit,
    "checksum": id_checksum,
    "vector_dim": vector_dim,
    "status": "INITIALIZED"
}
 
print("=== SYSTEM CONFIGURATION ===")
for key, value in sys_config.items():
    print(f"{key.upper()}: {value}")
 
def get_category(score):
    if score >= 90:
        return "EXCELLENT"
    elif score >= 80:
        return "VERY GOOD"
    elif score >= 75:
        return "GOOD"
    elif score >= 60:
        return "PASSING"
    else:
        return "NEEDS IMPROVEMENT"
```

```python
print("\n=== 1. GENERATED PERFORMANCE RECORDS ===")
 
students = [f"{LAST_NAME}_S{n}" for n in range(1, 6)]
subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]
 
artist_positions = [pos for pos in range(1, len(FAVORITE_ARTIST) + 1)]
print(f"Artist Positions: {artist_positions}")
 
# Score = 50 + ((position * seed + checksum) % 50), so scores stay between 50 and 99
scores = [50 + ((pos * sys_config["base_seed"] + sys_config["checksum"]) % 50) for pos in artist_positions]
print(f"Generated Scores: {scores}")
 
performance_records = []
 
for i in range(5):
    performance_records.append((students[i], subjects[i], scores[i]))
 
performance_records.append((students[0], subjects[0], scores[5]))
 
performance_records.append((students[1], subjects[4], scores[6]))
performance_records.append((students[4], subjects[3], scores[7]))
 
performance_records.append((students[2], subjects[4], None))
 
print(f"\n{'STUDENT':<12}{'SUBJECT':<14}{'SCORE'}")
for student, subject, score in performance_records:
    if score is None:
        print(f"{student:<12}{subject:<14}MISSING")
    else:
        print(f"{student:<12}{subject:<14}{score}")
```

```python
print("\n=== 2. ORGANIZED STUDENT/SUBJECT DATA ===")
 
student_data = {}
for student, subject, score in performance_records:
    if student not in student_data:
        student_data[student] = []
    student_data[student].append((subject, score))
 
print("Records per Student:")
for student, record_list in student_data.items():
    print(f"{student}: {record_list}")
 
student_scores = {}
subject_scores = {}
for student, subject, score in performance_records:
    if score is not None:
        if student not in student_scores:
            student_scores[student] = []
        student_scores[student].append(score)
 
        if subject not in subject_scores:
            subject_scores[subject] = []
        subject_scores[subject].append(score)
 
print("\nValid Scores per Subject:")
for subject, score_list in subject_scores.items():
    print(f"{subject}: {score_list}")
```

```python
print("\n=== 3. REPEATED/INCOMPLETE RECORDS ===")
 
seen_pairs = set()
repeated_records = []
incomplete_records = []
 
for student, subject, score in performance_records:
    pair = (student, subject)
    if pair in seen_pairs:
        repeated_records.append((student, subject, score))
    else:
        seen_pairs.add(pair)
 
    if score is None:
        incomplete_records.append((student, subject))
 
print(f"Repeated Records: {repeated_records}")
print(f"Incomplete Records: {incomplete_records}")
print(f"Total Records: {len(performance_records)}")
print(f"Valid Scores: {len(performance_records) - len(incomplete_records)}")
 
```

```python
print("\n=== 4. PERFORMANCE CATEGORIES ===")
 
categorized_records = []
category_counts = {}
 
for student, subject, score in performance_records:
    if score is not None:
        category = get_category(score)
        categorized_records.append((student, subject, score, category))
 
        if category not in category_counts:
            category_counts[category] = 0
        category_counts[category] += 1
 
print(f"{'STUDENT':<12}{'SUBJECT':<14}{'SCORE':<8}{'CATEGORY'}")
for student, subject, score, category in categorized_records:
    print(f"{student:<12}{subject:<14}{score:<8}{category}")
 
print("\nCategory Counts:")
for category, count in category_counts.items():
    print(f"{category}: {count}")
```

```python
print("\n=== 5. STUDENT AND SUBJECT SUMMARIES ===")
 
student_summary = {}
for student, score_list in student_scores.items():
    student_summary[student] = {
        "records": len(score_list),
        "highest": max(score_list),
        "lowest": min(score_list),
        "average": round(sum(score_list) / len(score_list), 2)
    }
 
print("Student Summary:")
for student, summary in student_summary.items():
    print(f"{student}: {summary}")
 
subject_summary = {}
for subject, score_list in subject_scores.items():
    subject_summary[subject] = {
        "records": len(score_list),
        "highest": max(score_list),
        "lowest": min(score_list),
        "average": round(sum(score_list) / len(score_list), 2)
    }
 
print("\nSubject Summary:")
for subject, summary in subject_summary.items():
    print(f"{subject}: {summary}")
```

```python
print("\n=== 6. ORDERED RESULTS ===")
 
student_ranking = []
for student, summary in student_summary.items():
    student_ranking.append((summary["average"], student))
student_ranking.sort(reverse=True)
 
print("Students by Average (Highest to Lowest):")
rank = 1
for average, student in student_ranking:
    print(f"{rank}. {student}: {average}")
    rank += 1
 
subject_ranking = []
for subject, summary in subject_summary.items():
    subject_ranking.append((summary["average"], subject))
subject_ranking.sort(reverse=True)
 
print("\nSubjects by Average (Highest to Lowest):")
rank = 1
for average, subject in subject_ranking:
    print(f"{rank}. {subject}: {average}")
    rank += 1
 
all_scores = [score for student, subject, score in performance_records if score is not None]
all_scores.sort(reverse=True)
print(f"\nAll Valid Scores (Highest to Lowest): {all_scores}")
```

```python
print("\n=== FINAL STUDENT PERFORMANCE REPORT ===")
print(f"Operator: {sys_config['operator']}")
print(f"Total Records: {len(performance_records)}")
print(f"Valid Scores: {len(all_scores)}")
print(f"Repeated Records: {len(repeated_records)}")
print(f"Incomplete Records: {len(incomplete_records)}")
print(f"Overall Average: {round(sum(all_scores) / len(all_scores), 2)}")
print(f"Top Student: {student_ranking[0][1]} ({student_ranking[0][0]})")
print(f"Lowest Student: {student_ranking[-1][1]} ({student_ranking[-1][0]})")
print(f"Best Subject: {subject_ranking[0][1]} ({subject_ranking[0][0]})")
print(f"Weakest Subject: {subject_ranking[-1][1]} ({subject_ranking[-1][0]})")
```
